```python
# Find numbers greater than 50 and store them in a new list.
# numbers = [20, 55, 30, 80, 45, 90, 10]

numbers = [20, 55, 30, 80, 45, 90, 10]

result = [n for n in numbers if n > 50]

print(result)
```

```python
# Find numbers between 20 and 50 and store them in a new list.
# numbers = [10, 25, 35, 60, 45, 15, 50, 70]

numbers = [10, 25, 35, 60, 45, 15, 50, 70]

result = [n for n in numbers if 20 <= n <= 50]

print(result)
```

```python
# Find numbers divisible by both 3 and 5 and store them in a new list.
# numbers = [10, 15, 20, 30, 45, 50, 60, 75]

numbers = [10, 15, 20, 30, 45, 50, 60, 75]

result = [
    n for n in numbers
    if n % 3 == 0 and n % 5 == 0
]

print(result)
```

```python
# Find names having more than 4 characters and store them in a new list.
# names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]

names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]

result = [name for name in names if len(name) > 4]

print(result)
```

```python
# Find names starting with "A" and store them in a new list.
# names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]

names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]

result = [name for name in names if name.startswith("A")]

print(result)
```

```python
# Extract vowels from a string and store the result in a new list.
# text = "programming"

text = "programming"

result = [ch for ch in text if ch in "aeiou"]

print(result)
```

```python
# Extract digits from a string and store the result in a new list.
# data = "Python123Programming456"

data = "Python123Programming456"

result = [ch for ch in data if ch.isdigit()]

print(result)
```

```python
# Replace every even number with "Even" and every odd number with "Odd", then store the result in a new list.
# numbers = [10, 15, 20, 25, 30]
# Expected Output:
# ['Even', 'Odd', 'Even', 'Odd', 'Even']

numbers = [10, 15, 20, 25, 30]

result = [
    "Even" if n % 2 == 0 else "Odd"
    for n in numbers
]

print(result)
```

```python
# Replace negative numbers with 0 and store the result in a new list.
# numbers = [10, -5, 20, -8, 30, -2]
# Expected Output:
# [10, 0, 20, 0, 30, 0]

numbers = [10, -5, 20, -8, 30, -2]

result = [n if n >= 0 else 0 for n in numbers]

print(result)
```

```python
# Give a 10% discount to products costing ₹1,000 or more; otherwise, give a 5% discount. Store the final prices in a new list.
# prices = [500, 1200, 800, 2000, 1500]
# Expected Output:
# [475.0, 1080.0, 760.0, 1800.0, 1350.0]

prices = [500, 1200, 800, 2000, 1500]

result = [
    p * 0.90 if p >= 1000 else p * 0.95
    for p in prices
]

print(result)
```

```python
# Find employees earning more than ₹50,000.
# employees = [
# {"name": "Amit", "salary": 45000},
# {"name": "Rahul", "salary": 60000},
# {"name": "Neha", "salary": 75000},
# {"name": "Rohit", "salary": 35000}
# ]
# Expected Output:
# ['Rahul', 'Neha']

employees = [
    {"name": "Amit", "salary": 45000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Neha", "salary": 75000},
    {"name": "Rohit", "salary": 35000}
]

result = [
    e["name"] for e in employees
    if e["salary"] > 50000
]

print(result)
```

```python
# Find products costing more than ₹1,000.
# prices = [499, 1299, 799, 2499, 999, 1599]
# Expected Output:
# [1299, 2499, 1599]

prices = [499, 1299, 799, 2499, 999, 1599]

result = [p for p in prices if p > 1000]

print(result)
```

```python
# Give a 20% discount only to products costing more than ₹2,000.
# prices = [500, 1200, 2500, 3000, 1500, 5000]
# Expected Output:
# [2000.0, 2400.0, 4000.0]

prices = [500, 1200, 2500, 3000, 1500, 5000]

result = [p * 0.80 for p in prices if p > 2000]

print(result)
```

```python
# Bank Transactions
# Positive values represent deposits. Find all deposits and all withdrawals.
# transactions = [5000, -2000, 3000, -1000, 7000, -500]
# Expected deposits:
# [5000, 3000, 7000]
# Expected withdrawals:
# [-2000, -1000, -500]

transactions = [5000, -2000, 3000, -1000, 7000, -500]

deposits = [t for t in transactions if t > 0]

withdrawals = [t for t in transactions if t < 0]

print("Deposits:", deposits)
print("Withdrawals:", withdrawals)
```

```python
# Find names of students who scored 60 or more.
# students = [
# {"name": "Amit", "marks": 75},
# {"name": "Rahul", "marks": 45},
# {"name": "Neha", "marks": 90},
# {"name": "Rohit", "marks": 55}
# ]
# Expected Output:
# ['Amit', 'Neha']

students = [
    {"name": "Amit", "marks": 75},
    {"name": "Rahul", "marks": 45},
    {"name": "Neha", "marks": 90},
    {"name": "Rohit", "marks": 55}
]

result = [
    s["name"] for s in students
    if s["marks"] >= 60
]

print(result)
```

```python
# Find names of employees from the IT department.
# employees = [
# {"name": "Amit", "department": "IT"},
# {"name": "Rahul", "department": "HR"},
# {"name": "Neha", "department": "IT"},
# {"name": "Rohit", "department": "Sales"}
# ]
# Expected Output:
# ['Amit', 'Neha']

employees = [
    {"name": "Amit", "department": "IT"},
    {"name": "Rahul", "department": "HR"},
    {"name": "Neha", "department": "IT"},
    {"name": "Rohit", "department": "Sales"}
]

result = [
    e["name"] for e in employees
    if e["department"] == "IT"
]

print(result)
```

```python
# Find employees who earn more than ₹60,000 and belong to IT.
# employees = [
# {"name": "Amit", "department": "IT", "salary": 55000},
# {"name": "Rahul", "department": "HR", "salary": 70000},
# {"name": "Neha", "department": "IT", "salary": 80000},
# {"name": "Rohit", "department": "IT", "salary": 45000}
# ]
# Expected Output:
# ['Neha']

employees = [
    {"name": "Amit", "department": "IT", "salary": 55000},
    {"name": "Rahul", "department": "HR", "salary": 70000},
    {"name": "Neha", "department": "IT", "salary": 80000},
    {"name": "Rohit", "department": "IT", "salary": 45000}
]

result = [
    e["name"] for e in employees
    if e["salary"] > 60000 and e["department"] == "IT"
]

print(result)
```
