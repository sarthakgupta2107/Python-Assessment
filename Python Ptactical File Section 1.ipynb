```python
#Print all even numbers from 1 to 20
for i in range(1, 21):
    if i % 2 == 0:
        print(i, end=" ")
```

```python
# Print all odd numbers from 1 to 20
for i in range(1, 21):
    if i % 2 != 0:
        print(i, end=" ")
```

```python
# Print multiples of 5 from 5 to 50
for i in range(5, 51, 5):
    print(i, end=" ")
```

```python
# Find the sum of numbers from 1 to 10
total = 0

for i in range(1, 11):
    total += i

print("Sum =", total)
```

```python
# Find the sum of even numbers from 1 to 100
total = 0

for i in range(2, 101, 2):
    total += i

print("Sum of even numbers =", total)
```

```python
# Find the sum of odd numbers from 1 to 100.
total=0
for i in range(1, 101, 2):
    total += i

print("Sum of odd numbers =", total)
```

```python
# Print the multiplication table of 7.
for i in range(1, 11):
    print("7 x", i, "=", 7 * i)
```

```python
# Take a number from the user and print its multiplication table.
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

```python
# Print numbers from 1 to 100 that are divisible by both 3 and 5.
for i in range(1, 101):
    if i % 3 == 0 and i % 5 == 0:
        print(i, end=" ")
```

```python
# Count even and odd numbers between 1 and 50.
even = 0
odd = 0

for i in range(1, 51):
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even numbers =", even)
print("Odd numbers =", odd)
```

```python
# Print the elements of a list in reverse order.
data = [22, 33, 12, 32, 45, 13]

for i in range(len(data) - 1, -1, -1):
    print(data[i], end=" ")
```

```python
# Find the largest number in a list using a for loop.
data = [22, 55, 12, 89, 34, 67]
largest = data[0]

for num in data:
    if num > largest:
        largest = num

print("Largest number =", largest)
```

```python
# Find the smallest number in a list.
data = [22, 55, 12, 89, 34, 67]
smallest = data[0]

for num in data:
    if num < smallest:
        smallest = num

print("Smallest number =", smallest)
```

```python
# Count vowels in a string.
text = input("Enter a string: ")
count = 0

for ch in text.lower():
    if ch in "aeiou":
        count += 1

print("Number of vowels =", count)
```

```python
# Count the number of spaces in a sentence.
text = "Python is easy to learn"
count = 0

for ch in text:
    if ch == " ":
        count += 1

print("Number of spaces =", count)
```
