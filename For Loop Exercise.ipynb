```python
# Initialize total sum variable
total_sum = 0

# Loop through numbers 1 to 10
for i in range(1, 11):
    total_sum += i

print("The sum is:", total_sum)
```

```python
# Initialize total sum variable
even_sum = 0

# Loop through even numbers from 2 to 100
for i in range(2, 101, 2):
    even_sum += i

print("The sum of even numbers is:", even_sum)
```
