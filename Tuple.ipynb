```python
# Standard tuple creation using parentheses
data1 = (10, 20, 30, 40, 50)
print("data1 type:", type(data1))  # Output: <class 'tuple'>

# Parentheses are optional; the comma is what matters
data2 = 10, 20, 30, 40, 50
print("data2 type:", type(data2))  # Output: <class 'tuple'>

# TRAP: A single number in parentheses is just an integer
data3 = (10)
print("data3 type:", type(data3))  # Output: <class 'int'>

# FIX: To make a single-element tuple, you MUST use a trailing comma
data4 = (10,)
print("data4 type:", type(data4))  # Output: <class 'tuple'>
```

```python
# Tuples are ordered and allow duplicate values
data = (10, 20, 30, 40, 50, 30)

# Positive indexing starts at 0 (left to right)
print("Element at index 2:", data[2])   # Output: 30

# Negative indexing starts at -1 (right to left)
print("Element at index -1:", data[-1]) # Output: 30
print("Full tuple with duplicates:", data)
```

```python
data = (10, 20, 30, 40, 50)

try:
    # Attempting to change an element will fail
    data[0] = 60
except TypeError as e:
    print(f"Error caught successfully: {e}") 
    # Output: 'tuple' object does not support item assignment
```

```python
# Tuple Packing: Storing multiple values into one variable
student_data = ("Amit", "MCA", 23, True)
print("Packed data type:", type(student_data))

# Tuple Unpacking: Extracting elements into separate variables
name, course, age, attendance = student_data
print(f"Unpacked -> Name: {name}, Course: {course}")

# Demonstrating ValueError (Variable mismatch)
try:
    # 4 elements in the tuple, but only providing 3 variables
    name, course, age = student_data
except ValueError as e:
    print(f"Error caught successfully: {e}") 
    # Output: too many values to unpack (expected 3)
```

```python
record = ("Amit", "MCA", 23, True)

# The * operator gathers all remaining elements into a list
name, *remaining_details = record

print("Name variable:", name)               # Output: Amit
print("Remaining details list:", remaining_details)  # Output: ['MCA', 23, True]
```

```python
# Define a collection of numbers
numbers = (5, 15, 25, 15, 35)

# Calculate various metrics using built-in functions
total_count = len(numbers)
occurrences = numbers.count(15)
total_sum = sum(numbers)
highest = max(numbers)
lowest = min(numbers)
first_position = numbers.index(25)

# Display the results
print("Total number of elements:", total_count)
print("Number of times 15 appears:", occurrences)
print("Sum of all numbers:", total_sum)
print("Largest number:", highest)
print("Smallest number:", lowest)
print("Index position of 25:", first_position)
```
