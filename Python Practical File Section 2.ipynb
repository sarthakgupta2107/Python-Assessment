```python
# Write a Python program for ATM withdrawal.
# Conditions:
# - Check whether the PIN is correct.
# - If the PIN is correct, check whether the withdrawal amount is a multiple of 100.
# - If it is, check whether the balance is sufficient.
# - If sufficient, withdraw the amount.
# - Otherwise, print "Insufficient Balance".
# - If the PIN is incorrect, print "Invalid PIN".

balance = 20000
pin = input("Enter Your Pin: ")

if pin == "12345":
    amount = int(input("Enter Your Amount do want to Withdraw: "))
    if amount % 100 == 0:
        if balance > amount:
            balance -= amount  # balance=balance-amount
            print("Withdraw Amount is: ", amount)
            print("Remaining Balance is: ", balance)
        else:
            print("Paisa Nahi hai")
    else:
        print("Invalid Amount")
else:
    print("Invalid Pin")
```

```python
# Student Result System
# Write a program that takes marks of 3 subjects.
# Conditions:
# - If every subject has marks >= 40, calculate the percentage.
# - If percentage >= 75, print "Distinction".
# - If percentage >= 60, print "First Division".
# - If percentage >= 50, print "Second Division".
# - Otherwise, print "Pass".
# - If any subject is below 40, print "Fail".

python = int(input("Enter Python Marks: "))
if python >= 40:
    java = int(input("Enter Java Marks: "))
    if java >= 40:
        dbms = int(input("Enter DBMS Marks: "))
        if dbms >= 40:
            totalMarks = python + java + dbms
            per = totalMarks / 3
            if per >= 75:
                print("Student Per is: ", per)
                print("Grade A+")
            elif per >= 60:
                print("Student Per is: ", per)
                print("Grade A")
            elif per >= 50:
                print("Student Per is: ", per)
                print("Grade B")
            else:
                print("Student Per is: ", per)
                print("Pass")
        else:
            print("Fail in DBMS")
    else:
        print("Fail in Java")
else:
    print("Fail in Python")
```

```python
# Bank Loan Eligibility
# A bank gives a loan based on the following conditions:
# - Age must be >= 21.
# - If age is valid, salary must be >= ₹25,000.
# - If salary is valid:
# - Credit score >= 750: Loan approved.
# - Credit score >= 650: Loan approved with higher interest.
# - Otherwise: Loan rejected.

age=int(input("Enter Your Age: "))
if(age>=21):
    salary=int(input("Enter Your Salary: "))
    if(salary>=25000):
        creditScore=int(input("Enter Your Credit Score: "))
        if(creditScore>=750):
            print("Loan Approved.")
            print("Intreast Rate: 8%")
        elif(creditScore>=650):
            print("Loan Approved.")
            print("Intreast Rate: 10%")
        else:
            print("Loan Not Approved")
    else:
        print("Salary Too Low")
else:
    print("Age is not Valid")
```

```python
# Login + Role-Based Access
# Create a login system.
# Conditions:
# - Check username.
# - If username is correct, check password.
# - If password is correct, check role:
# - "admin" -> Full Access
# - "teacher" -> Teacher Dashboard
# - "student" -> Student Dashboard
# - Anything else -> Invalid Role
# - Wrong password -> "Wrong Password"
# - Wrong username -> "Invalid Username"

userName=input("Enter User Name: ")
if(userName=="Admin" or userName=="admin"):
    password=input("Enter User Password: ")
    if(password=="12345"):
        role=input("Enter User Role: ")
        if(role=="Admin" or role=="admin"):
            print("Full Access")
        elif(role=="Trainer"):
            print("Trainer Dashboard")
        elif(role=="Student" or role=="student"):
            print("Student Dashboard")
        else:
            print("Invalid User Role")
    else:
        print("Wrong Password")
else:
    print("Invalid User Name")
```

```python
# Movie Ticket Booking
# Create a ticket booking system.
# Conditions:
# - Check whether seats are available.
# - If seats are available, check customer age:
# - Age < 5 -> Free
# - Age 5-12 -> 50% discount
# - Age > 60 -> 30% discount
# - Otherwise -> Full price
# - If no seats are available, print "House Full".

seats_available = int(input("Enter available seats: "))

if seats_available > 0:
    age = int(input("Enter customer age: "))
    price = 200

    if age < 5:
        final_price = 0
    elif age <= 12:
        final_price = price * 0.50
    elif age > 60:
        final_price = price * 0.70
    else:
        final_price = price

    print("Ticket price =", final_price)
else:
    print("House Full")
```

```python
# Electricity Bill
# Conditions:
# - 0-100 units -> ₹5/unit
# - 101-200 units -> ₹7/unit
# - 201-300 units -> ₹10/unit
# - Above 300 units -> ₹15/unit
# - Additionally, if bill > ₹3,000, add a 10% surcharge.

units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = units * 7
elif units <= 300:
    bill = units * 10
else:
    bill = units * 15

if bill > 3000:
    bill += bill * 0.10

print("Electricity bill =", bill)
```

```python
# Employee Bonus
# Conditions:
# - If experience >= 1 year and salary >= ₹50,000:
# - Performance >= 90 -> 20% bonus
# - Performance >= 75 -> 10% bonus
# - Otherwise -> 5% bonus
# - If salary < ₹50,000:
# - Performance >= 90 -> 15% bonus
# - Otherwise -> 5% bonus

experience = float(input("Experience in years: "))
salary = float(input("Salary: "))
performance = int(input("Performance score: "))

if experience >= 1 and salary >= 50000:
    if performance >= 90:
        rate = 0.20
    elif performance >= 75:
        rate = 0.10
    else:
        rate = 0.05

elif salary < 50000:
    if performance >= 90:
        rate = 0.15
    else:
        rate = 0.05
else:
    rate = 0

print("Bonus =", salary * rate)
```

```python
# Driving License
# Conditions:
# - Age >= 18.
# - Check whether a learner license is available.
# - Driving test score >= 60.

age = int(input("Enter age: "))

if age >= 18:
    learner = input("Do you have a learner license? (yes/no): ").lower()

    if learner == "yes":
        score = int(input("Enter driving test score: "))

        if score >= 60:
            print("License eligible")
        else:
            print("Driving test failed")
    else:
        print("Learner license required")
else:
    print("Age requirement not met")
```

```python
# ATM - Difficult Challenge
# Conditions:
# - PIN must be correct.
# - Account must be active.
# - Amount must be positive.
# - Amount must be a multiple of ₹100.
# - Balance must be sufficient.
# - After withdrawal, a minimum balance of ₹500 must remain.
# - Amount > ₹20,000 -> Extra verification.

correct_pin = 1234
balance = 30000

pin = int(input("Enter PIN: "))

if pin == correct_pin:
    active = input("Is account active? (yes/no): ").lower()

    if active == "yes":
        amount = int(input("Enter amount: "))

        if amount > 0:
            if amount % 100 == 0:
                if balance - amount >= 500:

                    if amount > 20000:
                        verify = input(
                            "Extra verification passed? (yes/no): "
                        ).lower()

                        if verify != "yes":
                            print("Verification failed")
                            raise SystemExit

                    balance -= amount
                    print("Withdrawal successful")
                    print("Remaining balance =", balance)

                else:
                    print("Minimum balance of 500 must remain")
            else:
                print("Amount must be a multiple of 100")
        else:
            print("Amount must be positive")
    else:
        print("Account inactive")
else:
    print("Invalid PIN")
```
