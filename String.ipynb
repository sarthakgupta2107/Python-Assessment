```python
# Basic variables
a = 15
b = 4

# Standard operators
print("Addition (a + b):", a + b)
print("Subtraction (a - b):", a - b)
print("Multiplication (a * b):", a * b)
print("Standard Division (a / b):", a / b)

# Special operators
print("Floor Division (15 // 4):", 15 // 4)   # Rounds down to nearest integer
print("Negative Floor Division (-15 // 4):", -15 // 4) # Rounds toward negative infinity
print("Exponentiation (5 ** 3):", 5 ** 3)     # 5 raised to the power of 3
```

```python
# Define string variables
greeting = "Hello"
name = "Alex"

# Combine strings using concatenation (+)
welcome_message = greeting + " " + name + "!"

print(welcome_message)
```

```python
# Initial values
score = 100
bonus = 20

print("Initial score:", score)

# Add and assign (score = score + bonus)
score += bonus
print("After += bonus:", score)

# Subtract and assign (score = score - 30)
score -= 30
print("After -= 30:", score)

# Multiply and assign (score = score * 2)
score *= 2
print("After *= 2:", score)

# Divide and assign (score = score / 5)
score /= 5
print("After /= 5:", score) # Notice it converts to a decimal (float)
```
